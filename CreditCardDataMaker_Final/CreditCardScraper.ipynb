````python
import requests
import xml.etree.ElementTree as ET
import pandas as pd
import re

xml = requests.get(
    "https://www.hdfc.bank.in/sitemap.xml"
).content

root = ET.fromstring(xml)

urls = []

for elem in root.iter():
    if elem.tag.endswith("loc"):
        urls.append(elem.text)

cards = []

seen = set()

pattern1 = re.compile(
    r"^https://www\.hdfc\.bank\.in/credit-cards/([^/?#]+)/?$"
)



exclude = {
    "benefits",
    "mitc",
    "card-payment",
    "activate-credit-card",
    "add-on-credit-card",
    "credit-card-balance-transfer",
    "credit-card-referral-program",
    "track-my-application",
    "credit-card-offer",
    "credit-card-rewards",
    "forex-cards",
    "fuel-credit-cards",
    "lounge-access-cards",
    "premium-credit-card",
    "purchase-credit-card",
    "rupay-credit-card",
    "credit-card",
    "quick-cash"
}

for url in urls:

    slug = None

    m = pattern1.match(url)
    
    if m:
        slug = m.group(1)

    
    if slug is None:
        continue

    if not slug.lower().endswith("credit-card"):
        continue

    if slug.lower() in exclude:
        continue

    if url in seen:
        continue

    seen.add(url)

    cards.append({
        "bank": "HDFC",
        "card_name": slug.replace("-", " ").title(),
        "slug": slug,
        "card_url": url
    })

df = pd.DataFrame(cards)

df.to_excel(
    "hdfc_credit_cards.xlsx",
    index=False
)

print(df)
print(f"\nFound {len(df)} cards")
````

````python
import requests
import xml.etree.ElementTree as ET
import pandas as pd
import re

# Axis sitemap
xml = requests.get(
    "https://www.axis.bank.in/sitemap-english.xml"
).content

root = ET.fromstring(xml)

urls = []

for elem in root.iter():
    if elem.tag.endswith("loc"):
        urls.append(elem.text.strip())

cards = []
seen = set()

consumer_pattern = re.compile(
    r"^https://www\.axis\.bank\.in/cards/credit-card/([^/?#]+)/?$"
)

exclude = {
    "airport-lounge-access-credit-card",
    "compare-credit-cards",
    "credit-card-against-fixed-deposit",
    "credit-card-for-housewife",
    "credit-card-referral-program",
    "instant-loan-on-credit-card",
    "lifestyle-credit-cards",
    "lifetime-free-credit-card",
    "pre-approved-credit-card",
    "rewards-credit-card",
    "rupay-credit-card",
    "shopping-credit-card",
    "travel-credit-card",
    "cashback-credit-card",
    "fuel-credit-card"
}

for url in urls:

    m = consumer_pattern.match(url)

    if not m:
        continue

    slug = m.group(1)

    # Skip nested pages
    if "/" in slug:
        continue

    if slug.lower() in exclude:
        continue

    if url in seen:
        continue

    seen.add(url)

    cards.append({
        "bank": "AXIS",
        "card_name": slug.replace("-", " ").title(),
        "slug": slug,
        "card_url": url
    })

df = pd.DataFrame(cards)

df = df.sort_values("card_name")

df.to_excel(
    "axis_credit_cards.xlsx",
    index=False
)

print(df)
print(f"\nFound {len(df)} Axis credit cards")
````

Helper functions

````python
def safe_name(name):

    return re.sub(
        r'[<>:"/\\|?*]',
        "_",
        str(name)
    )


def classify_pdf(filename):

    name = filename.lower()

    for word in IGNORE:

        if word in name:
            return "IGNORE"

    best_category = None
    best_score = 0

    for category, keywords in RULES.items():

        score = 0

        for keyword in keywords:

            if keyword.lower() in name:
                score += 1

        if score > best_score:

            best_score = score
            best_category = category

    if best_category is None:
        return "MISCELLANEOUS"

    return best_category
````

card processing

````python
def process_card(bank, card, url):

    print("\n" + "=" * 80)
    print(f"Processing: {bank} | {card}")
    print("=" * 80)

    card_dir = ROOT_DIR / bank / card

    card_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    response = requests.get(
        url,
        headers=HEADERS,
        timeout=30
    )

    html = response.text

    soup = BeautifulSoup(
        html,
        "html.parser"
    )

    pdf_links = set()

    # ==================================================
    # METHOD 1 : Extract PDFs from anchor tags
    # ==================================================

    anchor_count = 0

    for a in soup.find_all(
        "a",
        href=True
    ):

        href = a["href"]

        if ".pdf" in href.lower():

            anchor_count += 1

            pdf_links.add(
                urljoin(
                    url,
                    href
                )
            )

    # ==================================================
    # METHOD 2 : Extract PDFs hidden in HTML / JSON
    # ==================================================

    matches = re.findall(
        r'[^"\']+\.pdf(?:\?[^"\']*)?',
        html,
        flags=re.IGNORECASE
    )

    regex_count = len(matches)

    for m in matches:

        m = m.strip()

        if m.startswith("http"):

            pdf_links.add(m)

        else:

            pdf_links.add(
                urljoin(
                    url,
                    m
                )
            )

    print(
        f"Anchor PDFs: {anchor_count}"
    )

    print(
        f"Regex PDFs: {regex_count}"
    )

    print(
        f"Unique PDFs Found: {len(pdf_links)}"
    )

    # ==================================================
    # DOWNLOAD + CLASSIFY
    # ==================================================

    for pdf_url in sorted(pdf_links):

        try:

            filename = Path(
                urlparse(pdf_url).path
            ).name

            category = classify_pdf(
                filename
            )

            if category == "IGNORE":

                print(
                    f"Skipped [IGNORE] {filename}"
                )

                continue

            category_dir = (
                card_dir /
                category
            )

            category_dir.mkdir(
                exist_ok=True
            )

            filepath = (
                category_dir /
                filename
            )

            if not filepath.exists():

                print(
                    f"Downloading [{category}] "
                    f"{filename}"
                )

                pdf_response = requests.get(
                    pdf_url,
                    headers=HEADERS,
                    timeout=60,
                    allow_redirects=True
                )

                if (
                    pdf_response.status_code == 200
                    and
                    "pdf" in pdf_response.headers.get(
                        "content-type",
                        ""
                    ).lower()
                ):

                    with open(
                        filepath,
                        "wb"
                    ) as f:

                        f.write(
                            pdf_response.content
                        )

                    print(
                        f"Saved: {filepath}"
                    )

                else:

                    print(
                        f"Failed: "
                        f"{pdf_response.status_code}"
                    )

            else:

                print(
                    f"Exists: {filename}"
                )

            INVENTORY.append({
                "bank": bank,
                "card": card,
                "category": category,
                "filename": filename,
                "url": pdf_url,
                "local_path": str(filepath)
            })

        except Exception as e:

            print(
                f"PDF Error: {e}"
            )
````

Cutting each pdf into max 10 pages for context issues

````python
from pathlib import Path
from pypdf import PdfReader, PdfWriter

ROOT_FOLDER = r"per_card_data"  # top-level folder
MAX_PAGES = 8


def split_pdf(pdf_path, max_pages=8):
    reader = PdfReader(pdf_path)
    total_pages = len(reader.pages)

    if total_pages <= max_pages:
        return

    stem = pdf_path.stem
    parent = pdf_path.parent

    part_num = 1

    for start in range(0, total_pages, max_pages):
        writer = PdfWriter()

        end = min(start + max_pages, total_pages)

        for page_num in range(start, end):
            writer.add_page(reader.pages[page_num])

        output_file = parent / f"{stem}{part_num}.pdf"

        with open(output_file, "wb") as f:
            writer.write(f)

        part_num += 1

    # Optional: delete original PDF after splitting
    #pdf_path.unlink()


for pdf_file in Path(ROOT_FOLDER).rglob("*.pdf"):
    try:
        split_pdf(pdf_file, MAX_PAGES)
        print(f"Processed: {pdf_file}")
    except Exception as e:
        print(f"Failed: {pdf_file}")
        print(e)
````

````python
STOP_PATTERNS = [
    "related products",
    "learning hub",
    "need information on credit card services",
    "look through our knowledge section",
    "privacy policy",
    "disclaimer",
    "copyright"
]


def table_to_markdown(rows):

    if not rows:
        return ""

    max_cols = max(len(r) for r in rows)

    normalized = [
        r + [""] * (max_cols - len(r))
        for r in rows
    ]

    header = "| " + " | ".join(normalized[0]) + " |"

    separator = "| " + " | ".join(
        ["---"] * max_cols
    ) + " |"

    body = "\n".join(
        "| " + " | ".join(r) + " |"
        for r in normalized[1:]
    )

    return "\n".join([
        header,
        separator,
        body
    ])


def extract_page(url):

    try:

        r = requests.get(
            url,
            headers=HEADERS,
            timeout=30
        )

        if r.status_code != 200:
            return None

    except Exception as e:

        print(
            f"Failed: {url} -> {e}"
        )

        return None

    soup = BeautifulSoup(
        r.text,
        "html.parser"
    )

    for tag in soup([
        "script",
        "style",
        "header",
        "footer",
        "nav",
        "noscript",
        "svg",
        "iframe"
    ]):
        tag.decompose()

    texts = []
    seen = set()

    # --------------------
    # TABLES FIRST
    # --------------------

    for table in soup.find_all("table"):

        rows = []

        for tr in table.find_all("tr"):

            cols = [
                td.get_text(
                    " ",
                    strip=True
                )
                for td in tr.find_all(
                    ["th", "td"]
                )
            ]

            if cols:
                rows.append(cols)

        if rows:

            table_text = (
                "\n[TABLE]\n" +
                table_to_markdown(rows) +
                "\n[/TABLE]\n"
            )

            if table_text not in seen:

                texts.append(
                    table_text
                )

                seen.add(
                    table_text
                )

        table.decompose()

    # --------------------
    # NORMAL CONTENT
    # --------------------

    for tag in soup.find_all(
        [
            "h1", "h2", "h3",
            "h4", "h5", "h6",
            "p", "li"
        ]
    ):

        text = tag.get_text(
            separator=" ",
            strip=True
        )

        text = " ".join(
            text.split()
        )

        normalized = text.lower()

        if any(
            pattern in normalized
            for pattern in STOP_PATTERNS
        ):
            print(
                f"Stopping at section: {text}"
            )
            break

        if len(text) < 5:
            continue

        if text in seen:
            continue

        texts.append(text)

        seen.add(text)

    return texts
````

````python
"""
Per-card credit card knowledge extractor (Jupyter-friendly).

Folder layout assumed:

    per_card_data/
        AXIS/
            Airtel Axis Bank Credit Card/
                summary.txt                <- read first
                AXIS_Airtel_..._.pdf       <- (if a pdf sits directly here, read after txt)
                LOUNGE/*.pdf
                MISCELLANEOUS/*.pdf
                OFFERS/*.pdf
                TERMS/*.pdf
            Axis Bank Ace Credit Card/
                ...
        HDFC/...
        IDFC FIRST/...

Each leaf folder that contains files (txt and/or pdf subfolders) is treated
as ONE "card". Its outputs go to:

    card_attributes/<Card Name>.json   <- cumulative merged knowledge
    state.json                          <- global list of processed files

Processing order, per card:
    1. All .txt files directly inside the card folder.
    2. All .pdf files, recursively, inside the card folder (LOUNGE, OFFERS, ...).

Each call to process_next_file() handles exactly ONE file (one txt or one
pdf) and merges its extracted info into that card's JSON. Run the
auto-loop cell at the bottom to keep calling it every ~1.15 sec until
everything is processed.
"""

from __future__ import annotations

import json
import os
import time
from pathlib import Path

from dotenv import load_dotenv
from cerebras.cloud.sdk import Cerebras
from groq import Groq
from pypdf import PdfReader

# --------------------------------------------------------------------------- #
# Configuration
# --------------------------------------------------------------------------- #

load_dotenv()

ROOT_DIR = Path(r"per_card_data")  # top-level folder containing card subfolders
OUTPUT_DIR = Path("card_attributes")  # one json per card goes here
STATE_FILE = Path("state.json")       # tracks globally processed files

CEREBRASGOOD = "gpt-oss-120b"    
CEREBRAS_FALLBACK = "zai-glm-4.7"           # cerebras
GROQ_MODEL_NAME = "qwen/qwen3-32b"  # groq
GROQ_FALLBACK_MODEL_NAME = "llama-3.3-70b-versatile"  # groq
MAX_PROMPT_CHARS = 55_000

from openai import OpenAI

client = OpenAI(
  base_url = "https://api.llm7.io/v1/",     
api_key = os.getenv("LLM7_API_KEY")
)

OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

EMPTY_TEMPLATE: dict = {
    "card_name": None,
    "card_network": None,
    "card_type": None,
    "joining_fee": None,
    "annual_fee": None,
    "fee_waiver_condition": None,

    "welcome_benefits": [],
    "reward_structure": [],
    "reward_redemption_options": [],
    "reward_exclusions": [],
    "reward_caps": [],

    "cashback_benefits": [],
    "milestone_benefits": [],

    "lounge_access": [],
    "airport_services": [],
    "priority_pass_benefits": [],

    "travel_benefits": [],
    "hotel_benefits": [],
    "airline_benefits": [],
    "forex_benefits": [],

    "dining_benefits": [],
    "movie_benefits": [],
    "golf_benefits": [],
    "fuel_benefits": [],
    "shopping_benefits": [],
    "entertainment_benefits": [],

    "insurance_benefits": [],
    "protection_benefits": [],
    "concierge_benefits": [],

    "emi_benefits": [],
    "business_benefits": [],

    "domestic_lounge_visits_per_year": None,
    "international_lounge_visits_per_year": None,
    "guest_lounge_access": None,
    "reward_rate": None,
    "forex_markup": None,
    "fuel_surcharge_waiver": None,

    "eligibility": [],
    "income_requirements": [],
    "documents_required": [],

    "charges": [],
    "interest_rates": [],
    "fees_and_penalties": [],

    "spend_conditions": [],
    "benefit_conditions": [],
    "renewal_benefits": [],

    "partner_programs": [],
    "merchant_offers": [],
    "limited_period_offers": [],

    "important_terms": [],
    "excluded_categories": [],

    "source_files_used": [],
}

PROMPT_TEMPLATE = """You are an expert credit card analyst.

The document below may mention multiple credit cards. Extract information
ONLY for this specific card: "{card_name}".
Ignore and do not include any details belonging to other cards, even if
they appear in the same document.

Rules:
1. Return ONLY valid JSON.
2. Do NOT explain.
3. Do NOT think aloud.
4. Do NOT use markdown.
5. Every field must be populated only from the provided document, and only
   if it relates to "{card_name}".
6. If a value is not found for this card, use null or [].
7. Include exact numerical values wherever possible.

Return ONLY valid JSON matching exactly this schema (same keys, same types):

{schema}

CARD DATA:

{page_text}
"""


# --------------------------------------------------------------------------- #
# File discovery: build an ordered queue of (card_name, file_path)
# --------------------------------------------------------------------------- #

def is_card_folder(folder: Path) -> bool:

    if any(folder.glob("*.txt")):
        return True

    if any(folder.glob("*.pdf")):
        return True

    pdf_count = sum(1 for _ in folder.rglob("*.pdf"))

    return pdf_count > 0


def discover_card_folders(root: Path):

    cards = []

    for bank_dir in root.iterdir():

        if not bank_dir.is_dir():
            continue

        for card_dir in bank_dir.iterdir():

            if not card_dir.is_dir():
                continue

            if (
                any(card_dir.glob("*.txt"))
                or any(card_dir.rglob("*.pdf"))
            ):
                cards.append(card_dir)

    return sorted(cards)


def files_for_card(card_folder: Path) -> list[Path]:
    """Ordered list: txt files in card root first, then all pdfs recursively."""
    txt_files = sorted(card_folder.glob("*.txt"))
    pdf_files = sorted(card_folder.rglob("*.pdf"))
    return txt_files + pdf_files


def build_queue(root: Path):

    queue = []

    for bank_dir in sorted(root.iterdir()):

        if not bank_dir.is_dir():
            continue

        bank_name = bank_dir.name

        for card_folder in sorted(bank_dir.iterdir()):

            if not card_folder.is_dir():
                continue

            if (
                not any(card_folder.glob("*.txt"))
                and not any(card_folder.rglob("*.pdf"))
            ):
                continue

            card_name = card_folder.name

            for f in files_for_card(card_folder):

                queue.append(
                    (
                        bank_name,
                        card_name,
                        f
                    )
                )

    return queue


# --------------------------------------------------------------------------- #
# State / output helpers
# --------------------------------------------------------------------------- #

def load_state() -> dict:
    if STATE_FILE.exists():
        return json.loads(STATE_FILE.read_text(encoding="utf-8"))
    return {"processed": []}


def save_state(state: dict) -> None:
    STATE_FILE.write_text(json.dumps(state, indent=2), encoding="utf-8")

def card_output_path(
    bank_name: str,
    card_name: str
) -> Path:

    bank_folder = OUTPUT_DIR / bank_name

    bank_folder.mkdir(
        parents=True,
        exist_ok=True
    )

    return bank_folder / f"{card_name}.json"


def load_card_knowledge(
    bank_name: str,
    card_name: str
) -> dict:

    path = card_output_path(
        bank_name,
        card_name
    )

    if path.exists():
        try:
            return json.loads(
                path.read_text(
                    encoding="utf-8"
                )
            )
        except json.JSONDecodeError:
            print(
                f"Warning: {path} invalid JSON."
            )

    data = json.loads(
        json.dumps(EMPTY_TEMPLATE)
    )

    data["card_name"] = card_name

    return data


def save_card_knowledge(
    bank_name: str,
    card_name: str,
    data: dict
) -> None:

    path = card_output_path(
        bank_name,
        card_name
    )

    path.write_text(
        json.dumps(
            data,
            indent=2,
            ensure_ascii=False
        ),
        encoding="utf-8"
    )

# --------------------------------------------------------------------------- #
# Text extraction
# --------------------------------------------------------------------------- #

def extract_text(file_path: Path) -> str:
    if file_path.suffix.lower() == ".pdf":
        reader = PdfReader(str(file_path))
        return "\n".join((page.extract_text() or "") for page in reader.pages)
    # .txt or other plain text
    return file_path.read_text(encoding="utf-8", errors="ignore")


# --------------------------------------------------------------------------- #
# Merging
# --------------------------------------------------------------------------- #

def is_empty(value) -> bool:
    return value is None or value == "" or value == [] or value == {}


def merge_value(old, new):
    if isinstance(old, list) or isinstance(new, list):
        old_list = old if isinstance(old, list) else ([] if is_empty(old) else [old])
        new_list = new if isinstance(new, list) else ([] if is_empty(new) else [new])

        merged = list(old_list)
        for item in new_list:
            if item not in merged and not is_empty(item):
                merged.append(item)
        return merged

    if is_empty(old):
        return new
    return old


def merge_knowledge(existing: dict, new_data: dict, source_file: str) -> dict:
    merged = dict(existing)

    for key, template_value in EMPTY_TEMPLATE.items():
        old_value = merged.get(key, template_value)
        new_value = new_data.get(key, template_value)
        merged[key] = merge_value(old_value, new_value)

    sources = merged.get("source_files_used", [])
    if source_file not in sources:
        sources.append(source_file)
    merged["source_files_used"] = sources

    return merged


# --------------------------------------------------------------------------- #
# LLM call
# --------------------------------------------------------------------------- #

PROVIDERS = [
    "llm7",
    "cerebrasgood",
    "cerebrasfallback",
    "groq",
    "groq_fallback"
]

CURRENT_PROVIDER = 0    



def _stream_chat(client, model: str, prompt: str) -> str:
    stream = client.chat.completions.create(
        model=model,
        messages=[
            {
                "role": "user",
                "content": prompt + "\n\nReturn ONLY JSON. No markdown. No explanations.",
            }
        ],
        temperature=0,
        stream=True,
    )

    result_parts: list[str] = []
    for chunk in stream:
        if chunk.choices and chunk.choices[0].delta and chunk.choices[0].delta.content:
            text = chunk.choices[0].delta.content
            print(text, end="", flush=True)
            result_parts.append(text)

    raw = "".join(result_parts).strip()

    if raw.startswith("```"):
        raw = raw.strip("`")
        if raw.lower().startswith("json"):
            raw = raw[4:]
        raw = raw.strip()

    return raw

import json
import re

def parse_json_response(raw: str) -> dict:

    raw = raw.strip()

    # Remove reasoning tags
    raw = re.sub(
        r"<think>.*?</think>",
        "",
        raw,
        flags=re.DOTALL
    ).strip()

    # Remove markdown fences
    if raw.startswith("```"):

        raw = raw.split("```", 1)[1]

        if raw.startswith("json"):
            raw = raw[4:]

        raw = raw.rsplit("```", 1)[0]

        raw = raw.strip()

    # Extract first JSON object
    match = re.search(
        r"\{.*\}",
        raw,
        flags=re.DOTALL
    )

    if not match:
        raise ValueError(
            f"No JSON found.\nResponse:\n{raw[:2000]}"
        )

    return json.loads(match.group(0))

def call_llm(text: str, card_name: str) -> dict:

    global CURRENT_PROVIDER

    schema_str = json.dumps(EMPTY_TEMPLATE, indent=4)

    prompt = PROMPT_TEMPLATE.format(
        schema=schema_str,
        page_text=text[:MAX_PROMPT_CHARS],
        card_name=card_name
    )

    start_idx = CURRENT_PROVIDER

    CURRENT_PROVIDER = (CURRENT_PROVIDER + 1) % len(PROVIDERS)

    for offset in range(len(PROVIDERS)):

        idx = (start_idx + offset) % len(PROVIDERS)

        provider = PROVIDERS[idx]

        try:

            print(f"\n[status] trying {provider}")

            if provider == "llm7":

                completion = client.chat.completions.create(
                    model="default",
                    messages=[
                        {
                            "role": "user",
                            "content": prompt
                        }
                    ],
                    temperature=0,
                    max_tokens=16384,
                    stream=False
                )

                raw = completion.choices[0].message.content
                print("\nRAW RESPONSE:")
                print(repr(raw))
                return parse_json_response(raw)

            elif provider == "cerebrasgood":

                cerebras_client = Cerebras(
                    api_key=os.environ["CEREBRAS_API_KEY"]
                )

                raw = _stream_chat(
                    cerebras_client,
                    CEREBRASGOOD,
                    prompt
                )

            elif provider == "cerebrasfallback":

                cerebras_client = Cerebras(
                    api_key=os.environ["CEREBRAS_API_KEY"]
                )

                raw = _stream_chat(
                    cerebras_client,
                    CEREBRAS_FALLBACK,
                    prompt
                )

            elif provider == "groq":  # groq

                groq_client = Groq(
                    api_key=os.environ["GROQ_API_KEY"]
                )

                raw = _stream_chat(
                    groq_client,
                    GROQ_MODEL_NAME,
                    prompt
                )

        
            elif provider == "groq_fallback": 

                groq_client = Groq(
                    api_key=os.environ["GROQ_API_KEY"]
                )

                raw = _stream_chat(
                    groq_client,
                    GROQ_FALLBACK_MODEL_NAME,
                    prompt
                )

            result = parse_json_response(raw)

            print(
                f"[success] {provider} "
                f"-> next starts at "
                f"{PROVIDERS[CURRENT_PROVIDER]}"
            )

            return result

        except Exception as exc:

            print(
                f"[failed] {provider}: "
                f"{exc!r}"
            )

    print("\n[all providers failed]")

    return {}
    


    


    


# --------------------------------------------------------------------------- #
# Core: process a single file
# --------------------------------------------------------------------------- #

def process_next_file() -> bool:

    if not os.environ.get("CEREBRAS_API_KEY"):
        raise RuntimeError(
            "CEREBRAS_API_KEY not set"
        )

    queue = build_queue(ROOT_DIR)

    if not queue:
        print(
            f"No txt/pdf files found under "
            f"{ROOT_DIR}"
        )
        return False

    state = load_state()

    processed = set(
        state.get("processed", [])
    )

    for bank_name, card_name, file_path in queue:

        if str(file_path) in processed:
            continue

        print(
            f"[{bank_name}] "
            f"[{card_name}] "
            f"Processing: {file_path.name}"
        )

        text = extract_text(file_path)

        print(
            f"  extracted characters: "
            f"{len(text):,}"
        )

        if not text.strip():

            print(
                "  warning: no extractable text"
            )

            processed.add(str(file_path))

            state["processed"] = sorted(
                processed
            )

            save_state(state)

            return True

        new_data = call_llm(
            text,
            card_name
        )

        knowledge = load_card_knowledge(
            bank_name,
            card_name
        )

        merged = merge_knowledge(
            knowledge,
            new_data,
            str(file_path)
        )

        if not merged.get("card_name"):
            merged["card_name"] = card_name

        save_card_knowledge(
            bank_name,
            card_name,
            merged
        )

        processed.add(
            str(file_path)
        )

        state["processed"] = sorted(
            processed
        )

        save_state(state)

        print(
            f"\n-> updated "
            f"{card_output_path(bank_name, card_name)}"
        )

        return True

    print(
        "All files already processed."
    )

    return False


# --------------------------------------------------------------------------- #
# Auto-loop for notebook use
# --------------------------------------------------------------------------- #

def run_auto_loop(
    interval_sec: float = 1.15,
    error_wait_sec: float = 20.0,
    max_iterations: int | None = None,
):
    """
    Repeatedly calls process_next_file() every `interval_sec` seconds
    until everything is processed (or max_iterations reached, or
    interrupted with the notebook 'Stop' / KeyboardInterrupt).

    If process_next_file() raises an error (e.g. API/rate-limit issue),
    wait `error_wait_sec` seconds and retry the SAME file instead of
    moving on at the normal interval.
    """
    i = 0
    try:
        while True:
            if max_iterations is not None and i >= max_iterations:
                print("Reached max_iterations, stopping.")
                break

            try:
                did_work = process_next_file()
            except Exception as exc:
                print(f"\nError: {exc!r}")
                waited = 0.0
                while waited < error_wait_sec:
                    print("[status] in nothing")
                    chunk = min(5.0, error_wait_sec - waited)
                    time.sleep(chunk)
                    waited += chunk
                continue  # retry same file, don't advance i or use normal interval

            if not did_work:
                break

            i += 1
            time.sleep(interval_sec)
    except KeyboardInterrupt:
        print("Stopped by user.")


# --------------------------------------------------------------------------- #
# Run directly (call this from a notebook cell)
# --------------------------------------------------------------------------- #

def run_once():
    return process_next_file()
````

````python
run_auto_loop()
# run_once()
````

````python
import os
import json
from pathlib import Path

import pandas as pd
from dotenv import load_dotenv
from openai import OpenAI

# ==========================================================
# CONFIG
# ==========================================================

load_dotenv()

INPUT_DIR = Path(r"card_attributes")
OUTPUT_DIR = Path("final_decision")

OUTPUT_DIR.mkdir(
    parents=True,
    exist_ok=True
)

client = OpenAI(
    base_url="https://api.llm7.io/v1/",
    api_key=os.getenv("LLM7_API_KEY")
)

PROMPT = """
You are a credit-card data extraction and normalization engine.

Return EXACTLY ONE valid JSON object.

Do not return explanations.
Every field in the schema must be present.

Rules:

* Missing string → null
* Missing number → null
* Missing boolean → false
* Missing array → []
* Monetary values must be numeric INR values.
* Percentage values must be numeric percentages without "%" sign.
* Never return text in numeric fields.
* Never hallucinate benefits.
* Extract only verifiable information.
* All booleans must be true or false.
* Arrays must always be arrays.
* benefits must always be an array of objects.


Card_type(choose 1): CASHBACK REWARDS TRAVEL FUEL SHOPPING PREMIUM BUSINESS 
Card_tier(choose 1): Beginner Mid-range Premium Invite-only


Field Types:
{
"card_name": string,
"issuer": string,
"network": string|null,

"card_type": string,
"card_tier": string,

"invite_only": boolean,

"income_requirement": number|0,
"eligibility_type": string|null,
"target_customer": string|null,

"age_min": number|0,
"age_max": number|1000,

"unsalaried_eligible": boolean,
"self_employed_eligible": boolean,
"student_eligible": boolean,

"joining_fee": number|0,
"annual_fee": number|0,
"renewal_fee": number|0,

"fee_waiver_spend": number|0,

"reward_program_type": string|null,

"base_reward_rate": number|0,

"online_reward_rate": number|0,
"offline_reward_rate": number|0,
"dining_reward_rate": number|0,
"travel_reward_rate": number|0,
"fuel_reward_rate": number|0,
"grocery_reward_rate": number|0,
"utility_reward_rate": number|0,
"international_reward_rate": number|0,

"welcome_bonus_value": number|0,
"renewal_bonus_value": number|0,
"milestone_value_annual": number|0,

"domestic_lounge_visits": number|0,
"international_lounge_visits": number|0,
"guest_lounge_visits": number|0,

"priority_pass_included": boolean,

"forex_markup": number|null,

"airline_transfer_partner_count": number|null,
"hotel_transfer_partner_count": number|null,

"airport_concierge": boolean,

"hotel_membership_value": number|0,
"airline_membership_value": number|0,

"movie_benefit_value": number|0,
"golf_benefit_value": number|0,
"dining_program_value": number|0,

"air_accident_cover": boolean,
"travel_insurance_cover": boolean,
"purchase_protection": boolean,

"benefits": [
{
"company_name": string,
"benefit_category": string,
"benefit_type": string,
"value": number|0,
"value_unit": string,
"max_limit": number|0,
"limit_period": string|null,
"conditions": string|null
}
],

"supports_online": boolean,
"supports_offline": boolean,
"supports_dining": boolean,
"supports_travel": boolean,
"supports_fuel": boolean,
"supports_grocery": boolean,
"supports_utility": boolean,
"supports_forex": boolean,

"excluded_categories": [string],

"primary_use_case": string|null,
"secondary_use_cases": [string],

"best_for": [string],
"avoid_for": [string],

"reward_structure": string|null,

"cashback_benefits": [string],
"travel_benefits": [string],
"milestone_benefits": [string],
"welcome_benefits": [string],

"estimated_total_annual_value": number|null,
"estimated_net_annual_value": number|null
}


"""

# ==========================================================
# HELPERS
# ==========================================================

def normalize_card(card_json):

    response = client.chat.completions.create(
        model="default",
        temperature=0,
        messages=[
            {
                "role": "system",
                "content": PROMPT
            },
            {
                "role": "user",
                "content": json.dumps(
                    card_json,
                    ensure_ascii=False
                )
            }
        ]
    )

    raw = response.choices[0].message.content.strip()

    if raw.startswith("```"):

        raw = raw.split("```", 1)[1]

        if raw.startswith("json"):
            raw = raw[4:]

        raw = raw.rsplit("```", 1)[0]

    return json.loads(raw)


def rebuild_excel():

    rows = []

    for bank_folder in OUTPUT_DIR.iterdir():

        if not bank_folder.is_dir():
            continue

        for card_file in bank_folder.glob("*.json"):

            try:

                row = json.loads(
                    card_file.read_text(
                        encoding="utf-8"
                    )
                )

                for k, v in row.items():

                    if isinstance(
                        v,
                        (list, dict)
                    ):
                        row[k] = json.dumps(
                            v,
                            ensure_ascii=False
                        )

                rows.append(row)

            except Exception:
                pass

    if not rows:
        return

    df = pd.DataFrame(rows)

    df.to_excel(
        OUTPUT_DIR / "cards_master.xlsx",
        index=False
    )

    print(
        f"Excel updated with {len(df)} cards"
    )


# ==========================================================
# PROCESS ONE CARD
# ==========================================================

def process_next_card():

    for bank_folder in sorted(
        INPUT_DIR.iterdir()
    ):

        if not bank_folder.is_dir():
            continue

        output_bank = (
            OUTPUT_DIR /
            bank_folder.name
        )

        output_bank.mkdir(
            parents=True,
            exist_ok=True
        )

        for card_file in sorted(
            bank_folder.glob("*.json")
        ):

            output_file = (
                output_bank /
                card_file.name
            )

            # Resume support
            if output_file.exists():
                continue

            print(
                f"\n[{bank_folder.name}] "
                f"{card_file.name}"
            )

            raw_card = json.loads(
                card_file.read_text(
                    encoding="utf-8"
                )
            )

            normalized = normalize_card(
                raw_card
            )

            output_file.write_text(
                json.dumps(
                    normalized,
                    indent=2,
                    ensure_ascii=False
                ),
                encoding="utf-8"
            )

            rebuild_excel()

            print(
                f"Saved -> {output_file}"
            )

            return True

    print(
        "\nAll cards processed."
    )

    return False


# ==========================================================
# RUN LOOP
# ==========================================================

def run_loop():

    while True:

        try:

            did_work = process_next_card()

            if not did_work:
                break

        except Exception as e:

            print(
                f"\nERROR: {e}"
            )

            break

    rebuild_excel()

    print(
        "\nFinished."
    )
````
